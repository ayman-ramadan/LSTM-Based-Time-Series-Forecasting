```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('weatherHistory.csv')
```

```python
data['Formatted Date'] = pd.to_datetime(data['Formatted Date'])
```

```python
data.drop('Loud Cover', axis=1, inplace=True)
data.drop('Daily Summary', axis=1, inplace=True)
```

```python
data_Cleaned = data.copy()

data_Cleaned.drop('Precip Type', axis=1, inplace=True)
data_Cleaned.drop('Apparent Temperature (C)', axis=1, inplace=True)
data_Cleaned.drop('Wind Bearing (degrees)', axis=1, inplace=True)
data_Cleaned.drop('Visibility (km)', axis=1, inplace=True)
data_Cleaned.drop('Summary', axis=1, inplace=True)
```

```python
data_Cleaned.head()
```

```python
data_Cleaned.sort_values(by='Formatted Date', inplace=True)
```

```python
data_Cleaned.drop('Formatted Date', axis=1, inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

data_Cleaned['Pressure (millibars)'] = scaler.fit_transform(data_Cleaned[['Pressure (millibars)']])
data_Cleaned['Temperature (C)'] = scaler.fit_transform(data_Cleaned[['Temperature (C)']])
data_Cleaned['Wind Speed (km/h)'] = scaler.fit_transform(data_Cleaned[['Wind Speed (km/h)']])
```

```python
data_Cleaned.describe()
```

```python
def create_sequences(data, seq_length=10):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length, -1])
    return np.array(X), np.array(y)
```

```python
X, Y = create_sequences(data_Cleaned.values)
```

```python
X[0], Y[0]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42, shuffle=False)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

X_train = torch.tensor(X_train, dtype=torch.float32)
Y_train = torch.tensor(Y_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
Y_test = torch.tensor(Y_test, dtype=torch.float32)

batch_size = 32
train_dataset = TensorDataset(X_train, Y_train)
test_dataset = TensorDataset(X_test, Y_test)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
class WeatherRNN(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers):
        super(WeatherRNN, self).__init__()
        self.gru = nn.GRU(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        out = self.fc(out[:, -1, :])
        return out

model = WeatherRNN(input_size=4, hidden_size=64, num_layers=2)

loss_fn = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
# Training loop
epochs = 20
for epoch in range(epochs):
    model.train()
    total_loss = 0
    for batch_X, batch_y in train_loader:
        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = loss_fn(outputs, batch_y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    # Evaluation
    model.eval()
    y_pred, y_true = [], []
    test_loss = 0
    with torch.no_grad():
        for batch_X, batch_y in test_loader:
            outputs = model(batch_X)
            y_pred.extend(outputs.cpu().numpy())
            y_true.extend(batch_y.cpu().numpy())
            test_loss += loss_fn(outputs, batch_y).item()

    rmse = np.sqrt(np.mean((np.array(y_pred) - np.array(y_true))**2))

    print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {total_loss/len(train_loader):.4f}, Test Loss: {test_loss/len(test_loader):.4f}, RMSE: {rmse:.2f}')
```
